```python
#VectorStore的用法
from langchain_text_splitters import MarkdownHeaderTextSplitter

#准备数据
markdown_content = open("resources/评估.md", "r").read()

#切分文本
headers_to_split = [
    ("#","Header1"),
    ("##","Header2"),
    ("###","Header3"),
]
markdown_splitter = MarkdownHeaderTextSplitter(
    headers_to_split,
    strip_headers=False # 不移除标题
    )

#切分文档, 返回的是 list[Document]
docs = markdown_splitter.split_text(markdown_content)

#给vectorstore添加id
for i, doc in enumerate(docs,start=1):
    doc.id = f"doc_{i}"

#初始化VectorStore
from langchain_core.vectorstores import InMemoryVectorStore
from langchain_ollama import OllamaEmbeddings

#创建ollama模型
embeddings = OllamaEmbeddings(model="qwen3-embedding:0.6b",dimensions=1024)

#初始化向量数据库
vectorstore = InMemoryVectorStore(
    embedding=embeddings
)

# 保存文本到向量数据库,返回文档id
ids = vectorstore.add_documents(docs)
print(ids)

#删除旧文档
vectorstore.delete(["doc_1"])
print("旧文档已删除")

#检索相似度
#用户问题
query = "论语中教育的目的是什么"

#相似度检索，返回值是list[Document]
results = vectorstore.similarity_search(query, k=2)
for result in results:
    print(result.page_content)

#2.4.4.2 带相似度得分的检索
results = vectorstore.similarity_search_with_score(query, k=2)
for result in results:
    print(result[0].page_content)
    print(result[1])
    


```
