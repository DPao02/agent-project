```python
#使用Ollama模型进行Embedding

from langchain_ollama import OllamaEmbeddings
import numpy as np


#创建ollama模型
embeddings = OllamaEmbeddings(model="qwen3-embedding:0.6b",dimensions=1024)

#向量化单条文本
vector = embeddings.embed_query("我爱上班")

#向量化document
vectors = embeddings.embed_documents(["我爱吃饭", "我爱学习", "我爱工作"])

#定义计算余弦相似度的函数
def cosine_similarity(v1, v2):
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))

for i,v in enumerate(vectors):
    print(f"Vector {i+1}与vector的相似度: {cosine_similarity(vector, v)}")
```
